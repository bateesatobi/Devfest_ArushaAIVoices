# 01 · Hello Gemini in Swahili
**Time:** 10 min · **Goal:** make your first call and see how the model handles your language.

Run the cells from top to bottom. **Do not share your key and do not upload private data.** Free-tier content may be used by Google to improve its products.

Get a key: https://aistudio.google.com/apikey

```python
# Setup: install the Gemini SDK and load your free AI Studio key
!pip -q install google-genai
import os, getpass
try:
    from google.colab import userdata          # Colab: left menu > Secrets > add GOOGLE_API_KEY
    API_KEY = userdata.get("GOOGLE_API_KEY")
except Exception:
    API_KEY = os.environ.get("GOOGLE_API_KEY") or getpass.getpass("Paste your AI Studio key: ")
from google import genai
from google.genai import types
client = genai.Client(api_key=API_KEY)
# Model names retire quickly. This tries several free-tier models and uses the first that works.
CANDIDATE_MODELS = ["gemini-3.5-flash-lite", "gemini-3.1-flash-lite", "gemini-3.5-flash"]
def pick_model():
    for m in CANDIDATE_MODELS:
        try:
            client.models.generate_content(model=m, contents="ping")
            print("Using model:", m)
            return m
        except Exception as e:
            print("Skipping", m, "->", str(e)[:100])
    raise RuntimeError("No model worked. Check your key, account and the model list at https://ai.google.dev/gemini-api/docs/models")
MODEL = pick_model()
```

## Step 1 · A first question in Swahili

```python
r = client.models.generate_content(
    model=MODEL,
    contents="Eleza kwa Kiswahili kwa nini majani ya mahindi yanaweza kugeuka manjano. Jibu kwa sentensi tatu fupi.",
)
print(r.text)
```

## Step 2 · Give the model a role (system instruction)

```python
config = types.GenerateContentConfig(
    system_instruction=(
        "You are a friendly agricultural advisor for smallholder farmers in East Africa. "
        "Answer briefly and practically, in the language the farmer uses. "
        "If you are not sure, say so and suggest the local extension officer."
    )
)
r = client.models.generate_content(model=MODEL, contents="Mahindi yangu yana wadudu. Nifanye nini?", config=config)
print(r.text)
```

## Step 3 · Same question, different languages
Edit the list. Ask a speaker of each language to rate the answer.

```python
question = "How can I tell if my maize has too much water or too little nitrogen?"
for lang in ["Swahili", "Luganda", "Ateso", "Runyankore", "Japadhola"]:
    r = client.models.generate_content(model=MODEL, contents=f"Answer in {lang}, in two short sentences: {question}", config=config)
    print(f"--- {lang} ---\n{r.text}\n")
```
